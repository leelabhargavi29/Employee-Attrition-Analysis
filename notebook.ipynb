## 9. Data Preprocessing
## 9.1 Copy Dataset

```python
df1 = df.copy()
```

## 9.2 Convert Target Variable

```python
df1["Attrition"] = df1["Attrition"].map({"Yes":1, "No":0})
```

## 9.3 Check Target Values

```python
df1["Attrition"].value_counts()
```

## 9.4 Encode Categorical Columns

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
for column in df1.columns:
    if df1[column].dtype.name in ["object", "string", "str"]:
        df1[column] = le.fit_transform(df1[column])
```

## 9.5 Check Encoded Data

```python
df1.head()
```

## 10. Feature Selection

```python
x = df1.drop("Attrition",axis=1)
y = df1["Attrition"]
```

## 11. Train-Test Split 

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,
                                                 y,
                                                 test_size=0.2,
                                                 random_state=42)
```

## 12. Model Building

```python
from sklearn.ensemble import GradientBoostingClassifier
model = GradientBoostingClassifier(random_state=42)
model.fit(x_train, y_train)
```

## 13. Model Evaluation
## 13.1 Predictions

```python
y_pred = model.predict(x_test)
```

## 13.2 Accuracy

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

## 13.3 Confusion Matrix

```python
from sklearn.metrics import confusion_matrix
print(confusion_matrix(y_test, y_pred))
```

## 13.4 Classification Report

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```
